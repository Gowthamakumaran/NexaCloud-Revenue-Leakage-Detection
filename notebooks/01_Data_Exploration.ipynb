**Loading The Dataset**

```python
from pathlib import Path

# Locate the dataset folder
possible_paths = [
    Path.cwd() / "data" / "operational",
    Path.cwd().parent / "data" / "operational",
    Path.home() / "Downloads" / "NexaCloud_Revenue_Leakage_Dataset"
]

data_dir = next(
    (path for path in possible_paths if path.is_dir()),
    None
)

if data_dir is None:
    raise FileNotFoundError("Dataset folder not found.")



# Load datasets
customers = pd.read_csv(data_dir / "customers.csv")
contracts = pd.read_csv(data_dir / "contracts.csv")
invoices = pd.read_csv(data_dir / "invoices.csv")
payments = pd.read_csv(data_dir / "payments.csv")
refunds = pd.read_csv(data_dir / "refunds.csv")

print("All five datasets loaded successfully!")
```

```python
payment_failure_rate = (
    payments.assign(
        failed=payments['payment_status'].eq('FAILED')
        ).groupby('payment_method')['failed'].agg(['sum','count'])
)
payment_failure_rate['failure_rate'] = (
    payment_failure_rate['sum']/
    payment_failure_rate['count']*100
)

payment_failure_rate.sort_values('failure_rate', ascending = False )
```

```python
pd.crosstab(payments['payment_status'],
            payments['failure_reason'],
            dropna=False
           )
```

```python
refund_customer = (
    refunds
    .merge(
        payments[['payment_id', 'invoice_id']],
        on='payment_id',
        how='left'
    )
    .merge(
        invoices[['invoice_id', 'contract_id']],
         on = 'invoice_id',   
         how = 'left'
    )
    .merge(
        contracts[['contract_id', 'customer_id']],
        on='contract_id',
        how= 'left'
    )
    .merge(
        customers[['customer_id', 'company_name']],
        on = 'customer_id',
        how = 'left'
    )
)   
```

```python
top_refund_customers = (
    refund_customer.groupby(['customer_id', 'company_name'])['refund_amount'].agg(['count','sum','mean']).sort_values('sum', ascending = False).head(10))
top_refund_customers
```
